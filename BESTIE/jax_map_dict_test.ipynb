```python
import jax
```

```python
d = {}
key = jax.random.key(123)
for i in range(100):
    d[f"dataset_{i}"] = {}
    #d[f"dataset_{i}"]["hist"] = f"hist_{i%5}"
    d[f"dataset_{i}"]["data"] = jax.random.uniform(key,(100,2,),dtype=jax.numpy.float64)
    key, _ = jax.random.split(key)
```

```python
def func(data):
    return data ** 2
```

```python
output = jax.tree_util.tree_map(lambda v: func(v),d)
```
